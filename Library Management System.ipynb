```python
class Book:
  def __init__(self,title,author):
    self.title = title
    self.author = author
    self.available = True

  def issue_book(self):
    if self.available:
      self.available = False
      print(f"{self.title} is issued")
    else:
      print(f"{self.title} is already been issued")

  def return_book(self):
    if self.available:
      print(f"{self.title} was not issued")

    else:
      self.available = True
      print(f"{self.title} has bees returned")

  def access_book(self):
    print("Accesiing book")

class Ebook(Book):
  def access_book(self):
    print(f"You can read {self.title} digitally ")

class PrintedBook(Book):
  def access_book(self):
    print(f"you can read {self.title} manually ")

class AudioBook(Book):
  def access_book(self):
    print(f"You can listen to {self.title}")


search = input("Enter thr book you want: ")
books = [
    Ebook("Pyhton Programming"," Guido van Rossum"),
    PrintedBook("Peer e kamil","Umaira Ahmad"),
    AudioBook("Crime and punishment","Fyodoor Dostoveskey")
]

for book in books:
  if search.lower() in book.title.lower():
    print("Book Found",f"Title: {book.title}")
    print(f"Author {book.author}")
    if book.available:
      print("Status = Available")
    else:
      print("Status = Issued")
    book.access_book()
    book.issue_book()
    break
else:
  print("Book was not found")

search = input("Enter the book you want to return: ")
for book in books:
  if search.lower() in book.title.lower():
    print("Book Found ",book.title)
    book.return_book()
    break
else:
  print("Book was not found bcuz The book is already in the record")
```
